## Podmínky ukončení předmětu – pracovní zadání k ověření se sylabem

1. Připravit odborný rukopis vhodný pro časopis indexovaný ve vhodné oborové databázi (např. Web of Science nebo Scopus); **příprava rukopisu není totéž co přijetí či publikace**. Rozsah a forma mají být potvrzeny vyučujícím v souladu s akreditovanými podmínkami předmětu.
2. Aktivně vystoupit na odborném semináři. Příklady témat:
   - Analýza podmínek **aktuální** grantové výzvy, způsobilost nákladů a hodnoticí kritéria.
   - Rozbor metodologie aktuálního oborového článku; diskuse validity, reprodukovatelnosti a omezení.
   - Vědecká komunikace, publikační etika, otevřená věda, autorské právo, duševní vlastnictví a práce s AI.

## Doktorský studijní program – § 47 zákona č. 111/1998 Sb.

1. Doktorské studium je zaměřeno na vědecké bádání a samostatnou tvůrčí činnost ve výzkumu nebo vývoji, případně uměleckou tvůrčí činnost (§ 47 odst. 1).
2. Standardní doba studia je zpravidla tři až čtyři roky v rozsahu stanoveném zákonem a akreditací (§ 47 odst. 2).
3. Studium probíhá podle individuálního studijního plánu pod vedením školitele (§ 47 odst. 3).
4. **Nový režim:** řádné ukončení státní závěrečnou zkouškou spočívající v obhajobě disertační práce (§ 47 odst. 4, řád UO čl. 40).
5. **Přechodný režim:** podle čl. II bodu 8 zákona č. 52/2025 Sb. a čl. 39 odst. 1 řádu UO studenti, jejichž první období výuky v daném doktorském studiu začalo **před 1. 3. 2025**, nadále konají **státní doktorskou zkoušku a obhajobu disertační práce**. Nelze tedy tvrdit, že samostatná doktorská zkouška již pro žádného studenta neexistuje.
6. Disertační práce musí splňovat zákonné a programové požadavky na původní výsledky tvůrčí činnosti. Vedle publikací mohou být podle právní úpravy relevantní též některé aplikační výsledky.
7. Absolvent získává titul **Ph.D.**; studium sleduje a hodnotí oborová rada (§ 47 odst. 5–6).

> Právní režim doktoranda se určuje podle skutečného začátku **prvního období výuky**, nikoli mechanicky podle data obhajoby, zápisu či aktuálního kalendářního roku. V hraničních případech je třeba vyhodnotit zákon č. 52/2025 Sb., akreditaci a studijní dokumentaci.

## Studijní a zkušební řád Univerzity obrany – pravidla doktorského studia

### Článek 13 – individuální studijní plán

- Pro každého doktoranda stanovuje individuální časový průběh a další podmínky studia.
- Zahrnuje studijní požadavky, tvůrčí povinnosti, plánované či doporučené zahraniční pobyty, stáže a pedagogické aktivity.
- Odpovídá za něj školitel a předloží jej oborové radě **do tří měsíců po zápisu do studia**.
- Plán a jeho změny schvaluje uskutečňovatel programu po projednání v oborové radě (čl. 13 odst. 2).

### Článek 39 – starší režim doktorského studia

Platí pro studenty, jejichž první období výuky začalo **před 1. březnem 2025** (odst. 1; odkaz na čl. II bod 8 zákona č. 52/2025 Sb.).

- Oddělená **státní doktorská zkouška a obhajoba disertační práce**.
- Nejméně pětičlenná komise; školitel ani školitel-specialista nesmí být členem komise pro obhajobu (odst. 5).
- Zkouška a obhajoba jsou veřejné, s výjimkami souvisejícími s utajovanými informacemi (odst. 8).
- Nejvýše dva pokusy (jedno opakování) pro státní doktorskou zkoušku nebo obhajobu; termíny a další postup v odst. 14–16.

### Článek 40 – nový režim doktorského studia

- **Státní závěrečná zkouška spočívající v obhajobě disertační práce**.
- Student podá žádost; musí odevzdat disertaci a splnit stanovené studijní povinnosti (odst. 2–3).
- Nejméně pětičlenná komise. Jejími členy nemohou být školitel ani školitel-specialista (odst. 4).
- Termín, místo a jazyk se stanoví do dvou měsíců od žádosti tak, aby se zkouška konala nejpozději do tří měsíců; vyrozumění nejméně deset pracovních dnů předem (odst. 6).
- Zkouška je veřejná, s omezeními pro utajované informace (odst. 7).
- Usnášeníschopnost vyžaduje účast nejméně dvou třetin členů, včetně předsedy nebo místopředsedy; hodnocení se přijímá nadpoloviční většinou **všech členů** (odst. 8).
- Hodnocení **prospěl(a) / neprospěl(a)**, nejvýše dva pokusy, tedy jedna oprava (odst. 10, 13).

### Článek 41 a příloha č. 2 – závěrečná práce

- Téma disertace schvaluje **oborová rada** (čl. 41 odst. 1).
- Text práce podléhá porovnání s databázemi pro kontrolu původnosti, se zvláštními pravidly pro chráněné informace (čl. 41 odst. 4).
- Podstatné požadavky (původnost, obsah, jazyk, rozsah, podoba) rozvádí příloha č. 2.

### Články 52–53 – přechod a účinnost

- Přechodná ustanovení čl. 52 upravují mj. již přerušené studium a dříve zahájená řízení.
- Řád je účinný od **1. října 2026**, kromě čl. 21 s dřívější účinností podle registrace (čl. 53 odst. 3).
- Pro rozdělení starého a nového doktorského režimu je rozhodující **speciální pravidlo čl. 39 odst. 1** v návaznosti na zákon č. 52/2025 Sb.; čl. 52 jej nenahrazuje.

## Instituce a role

- **Garant doktorského studijního programu:** akademický pracovník odpovědný za odborné a obsahové zajištění studijního programu podle příslušných právních a akreditačních pravidel; je předsedou oborové rady (§ 47 odst. 6).
- **Oborová rada:** sleduje a hodnotí průběh doktorského studia; její konkrétní pravomoci vyplývají z právních a vnitřních předpisů.
- **Školitel:** odborně vede doktoranda a odpovídá za přípravu individuálního studijního plánu v rozsahu čl. 13 řádu UO.
- **Národní akreditační úřad pro terciární vzdělávání:** současné označení akreditačního orgánu; podrobnosti jeho pravomocí a institucionální akreditace je vhodné probírat v samostatné části.
